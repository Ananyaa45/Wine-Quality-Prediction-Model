```python
df.columns=['fixed acidity','volatile acidity','citric acid','residual sugar','chlorides','free sulfur dioxide','total sulfur dioxide','density','pH','sulfates','alcohol','quality']
print('Data description:')
print(df.describe())
```

Q1.What is the most frequently occuring wine quality?What is the highest number in and the lowest number in the quanlity column?

```python
quality_counts = df['quality'].value_counts()
most_frequent_quality = quality_counts.idxmax()
highest_number = df['quality'].max()
lowest_number = df['quality'].min()
print("Most frequently occuring wine quality:", most_frequent_quality)
print("Highest number in the 'quality' column:", highest_number)
print("Lowest number in the 'quality' column:", lowest_number)
```

Q2.How is 'fixed acidity' correlated to the quality of wine?How does the alcohol content affect the 'quality'?How is the 'free Sulphur dioxide' content correlated to the quality of wine?

```python
correlation_matrix = df.corr()
print("Correlation matrix:")
print(correlation_matrix)
```

```python
fixedacidity_quality_correlation = df['fixed acidity'].corr(df['quality'])
print("Correlation between 'fixed acidity' and 'quality':", fixedacidity_quality_correlation)
```

```python
alcohol_quality_correlation = df['alcohol'].corr(df['quality'])
print("Correlation between 'alcohol' and 'quality':", alcohol_quality_correlation)
#
```

```python
freesulfurdioxide_quality_correlation = df['free sulfur dioxide'].corr(df['quality'])
print("Correlation between 'free sulfur dioxide' and 'quality':", freesulfurdioxide_quality_correlation)
```

Among these three(fixed acidity,alcohol,free sulphur dioxide)alcohol is most correlated with wine quality.Free sulfur dioxide has a negative correlation such that if free sulphur dioxide will increase wine quality will decrease.Alcohol has a high impact and fixed acidity has a low but positive impact.

Q3.	What is the average 'residual sugar' for the best quality wine and the lowest quality wine in the dataset?

```python
avg_residual_sugar_best_quality = df[df['quality'] == df['quality'].max()]['residual sugar'].mean()
avg_residual_sugar_lowest_quality = df[df['quality'] == df['quality'].min()]['residual sugar'].mean()
print("Average 'residual sugar' for the best quality wine:", avg_residual_sugar_best_quality)
print("Average 'residual sugar' for the lowest quality wine:", avg_residual_sugar_lowest_quality)
```

Q4.Does volatile acidity has an effect over the quality of wine samples in the dataset?

```python
volatile_acidity_quality_correlation = df['volatile acidity'].corr(df['quality'])
print("Correlation between 'volatile acidity' and 'quality':", volatile_acidity_quality_correlation)
```

Volatile acidity has a negative correlation with the quality of the wine.So when volatile acidity increases quality of the wine decreases.

Q5.	Train a Decision Tree model and Random Forest Model separately to predict the Quality of the given samples of wine. Compare the Accuracy scores for both models.



```python
X=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.009,random_state=0)
print(X_train)
print(X_test)
print(y_train)
print(y_test)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
classifier=DecisionTreeClassifier(criterion='entropy',random_state=0)
classifier.fit(X_train,y_train)
y_pred=classifier.predict(X_test)#predicting results on the basis of decision tree model
print(np.concatenate((y_pred.reshape(len(y_pred),1),y_test.reshape(len(y_test),1)),1))
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
accuracy_score(y_test,y_pred)
print(accuracy_score(y_test,y_pred))
```

Visualizing the decision tree model

```python
feature_names=['fixed acidity','volatile acidity','citric acid','residual sugar','chlorides','free sulfur dioxide','total sulfur dioxide','density','pH','sulfates','alcohol']
```

```python
class_names=sorted(np.unique(y_train).astype(str))
#Visualize existing Decision tree
print("Displaying decision tree")
plt.figure(figsize=(25,12))
from sklearn.tree import plot_tree
plot_tree(classifier,filled=True,feature_names=feature_names,class_names=class_names,rounded=True,fontsize=10)
plt.title("Decision Tree")
plt.show()
```

Random Forest Model

```python
rf_classifier=RandomForestClassifier(n_estimators=200,criterion='entropy',random_state=0)
rf_classifier.fit(X_train,y_train)
rf_y_pred=rf_classifier.predict(X_test)
print(np.concatenate((rf_y_pred.reshape(len(rf_y_pred),1),y_test.reshape(len(y_test),1)),1))#showing predicted and actual values side by side
```

```python
cm1=confusion_matrix(y_test,rf_y_pred)
print("confusion matrix is:")
print(cm1)
accuracy_score(y_test,rf_y_pred)
print("accuracy score is:")
print(accuracy_score(y_test,rf_y_pred))
```

Visualizing one tree from the Random Forest Model

```python
plt.figure(figsize=(25,12))
plot_tree(rf_classifier.estimators_[0],filled=True,feature_names=feature_names,class_names=class_names,rounded=True,fontsize=10)#first tree
plt.title("Random Forest Tree")
plt.show()
```

Accuracy of decision tree and one tree of random forest model is  equal.
